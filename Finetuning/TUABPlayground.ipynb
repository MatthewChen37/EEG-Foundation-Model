```python
import numpy as np
from TUABLoader import TUABLoader
import os
import torch
```

```python
root = "/home/azureuser/mycontainer/TUAB/v3.0.1/edf/processed"
```

```python
file_it = os.scandir(os.path.join(root, "train"))
train_files = []
```

```python
for i in range(300):
	entry = next(file_it)
	path = entry.path
	file_name = path.split("/")[-1]
	train_files.append(file_name)
np.random.shuffle(train_files)
```

```python
print(train_files)
```

```python
train_loader = torch.utils.data.DataLoader(
	TUABLoader(os.path.join(root, "train"),
				train_files, 200),
	batch_size=4,
	shuffle=True,
	drop_last=True,
	num_workers=4,
	persistent_workers=True,
)
```

```python
for idx, (x, y) in enumerate(train_loader):
	print(f"Batch {idx}, X shape: {x.shape}, Y shape: {y.shape}, Y: {y}")
	if idx == 10:
		break
```
